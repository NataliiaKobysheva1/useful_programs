Suma **n** początkowych wyrazów ciągu geometrycznego $ S_{n} $ zależy od wartości ilorazu **q**:

* Dla $q \neq 1$:
$$S_n = a_1 \cdot \frac{1 - q^n}{1 - q}$$

* Dla $q = 1$:
$$S_n = n \cdot a_1$$

```python
import matplotlib.pyplot as plt

# Przygotowanie danych (k od 1 do 20)
k_values = list(range(1, 21))
y_values = [k**2 for k in k_values]

# Tworzenie wykresu
plt.figure(figsize=(10, 6))

# Rysowanie punktów (wykres punktowy idealny dla wartości dyskretnych)
plt.scatter(k_values, y_values, color='blue', edgecolor='black', s=50, zorder=3, label='$y = k^2$')

# Opcjonalnie: delikatna linia pomocnicza trendu
plt.plot(k_values, y_values, color='lightblue', linestyle='--', alpha=0.6, zorder=2)

# Tytuł i opisy osi
plt.title('Wykres funkcji $y = k^2$ dla $k \in \{1, ..., 20\}$', fontsize=14, pad=15)
plt.xlabel('Wartość k', fontsize=12)
plt.ylabel('Wartość y', fontsize=12)

# Konfiguracja osi X, aby pokazywała każdą liczbę całkowitą od 1 do 20
plt.xticks(range(1, 21))

# Dodanie siatki pomocniczej
plt.grid(True, linestyle=':', alpha=0.6, zorder=1)

# Dodanie legendy
plt.legend(fontsize=12)

# 1. ZAPISYWANIE PLIKU (musi być PRZED plt.show())
plt.savefig('wykres.png', bbox_inches='tight', dpi=300)

# 2. Wyświetlenie wykresu w Colabie
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML
from base64 import b64encode

# Konfiguracja wykresu i sceny
fig, ax = plt.subplots(figsize=(5, 5))
ax.set_xlim(-1.2, 1.2)
ax.set_ylim(-1.2, 0.2)
ax.set_aspect('equal')
ax.grid(True, linestyle='--')

# Elementy wizualne wahadła (linia i punkt)
line, = ax.plot([], [], 'o-', lw=3, color='crimson', markersize=10)

# Dane fizyczne (ruch harmoniczny dla uproszczenia animacji)
frames = 60
theta = 0.5 * np.sin(np.linspace(0, 2 * np.pi, frames)) # Kąt w czasie

def init():
    line.set_data([], [])
    return line,

def animate(i):
    # Obliczenie współrzędnych końca wahadła (L=1)
    x = np.sin(theta[i])
    y = -np.cos(theta[i])
    line.set_data([0, x], [0, y])
    return line,

# Tworzenie animacji
ani = animation.FuncAnimation(fig, animate, init_func=init, frames=frames, interval=50, blit=True)
plt.close() # Zamknięcie statycznego okna wykresu

# Zapis do pliku video.mp4
ani.save('video.mp4', writer='ffmpeg', fps=20)

# Osadzenie i wyświetlenie wideo w Google Colab za pomocą HTML i Base64
mp4 = open('video.mp4', 'rb').read()
data_url = "data:video/mp4;base64," + b64encode(mp4).decode()
HTML(f'<video width="400" controls><source src="{data_url}" type="video/mp4"></video>')
```

```python
# Correcting the label for rendering issue in matplotlib mathtext
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Zmiana parametrów: Generujemy inną krzywą parametryczną 3D (Helisę stożkową / Conical Helix)
t = np.linspace(0, 10 * np.pi, 1000)
x = t * np.cos(t)
y = t * np.sin(t)
z = t

# # Alternatywna prosta helisa (jeśli wolisz zwykłą sprężynę, odkomentuj poniższe linie):
# x = np.cos(t)
# y = np.sin(t)
# z = t

# 3D Plot
fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(111, projection='3d')
ax.plot(x, y, z, label="New Parametric Curve", color='crimson', lw=2)
ax.set_title("3D Parametric Curve (Modified)", fontsize=14)
ax.set_xlabel("x(t)")
ax.set_ylabel("y(t)")
ax.set_zlabel("z(t)")
ax.legend()

# ZAPIS WYKRESU DO PLIKU (musi być wykonany PRZED plt.show())
plt.savefig('parametric_curve.png', bbox_inches='tight', dpi=300)

# Wyświetlenie wykresu w notebooku
plt.show()

```
